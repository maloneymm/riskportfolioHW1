```python
import pandas as pd
path = "/Users/meganmaloney/Documents/Risk HW1/riskportfolioHW1/multi_asset_etf_data.xlsx"
path = "./multi_asset_etf_data.xlsx" #added so it can run on other people's computers 
hw1_data = pd.read_excel(path, sheet_name = "excess returns", index_col = 0)
hw1_data = hw1_data.drop(columns = "QAI")
```

1. Summary Statistics

```python
mean = hw1_data.mean() * 12
vol = hw1_data.std() * (12 ** 0.5)
sharpe = mean / vol

summary_stats = pd.DataFrame({"Mean": mean, "Vol": vol, "Sharpe Ratio": sharpe})
summary_stats = summary_stats.sort_values(by = "Sharpe Ratio", ascending = False)
summary_stats
```

2. Descriptive Analysis

```python
corr_matrix = hw1_data.corr()
corr_matrix
```

```python
pairs = corr_matrix.unstack()
#unstack() "flattens" the 10x10 table into a long list of all 100 entries, each
#labeled with two tickers and their correlation value
pairs = pairs[pairs < 1]
#only want to look at the off-diagonal pairs (the diagonals are self-correlations)
print("Highest:", pairs.idxmax(), round(pairs.max(), 2))
print("Lowest:", pairs.idxmin(), round(pairs.min(), 2))
summary_stats.loc[["TIP", "IEF", "BWX"]]
```

- Listed private equity is US-heavy equity, so highly correlated with the S&P.

- Commodities against Treasuries has the lowest correlation

- TIPS have outperformed both domestic (IEF) and foreign bonds (BWX). 
Higher mean, lower volatility, and higher Sharpe than both IEF and BWX.
